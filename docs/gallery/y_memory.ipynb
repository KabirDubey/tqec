```python
from tqec.computation.block_graph import BlockGraph
from tqec.computation.cube import YHalfCube
from tqec.computation.pipe import PipeKind
from tqec.utils.position import Position3D
from tqec import compile_block_graph
```

```python
# Create a Y memory experiment manually - BlockGraph with a Y half cube
graph = BlockGraph()

y_init = graph.add_cube(Position3D(0, 0, 0), YHalfCube())
y_meas = graph.add_cube(Position3D(0, 0, 1), YHalfCube())

# Connect them with a timelike pipe (Z direction)
# You can choose either "zxo" or "xzo" for the pipe kind
kind = PipeKind.from_str("zxo")
pipe = graph.add_pipe(y_init, y_meas, kind)

print("Created Y memory block graph:")
print(f"Number of cubes: {len(graph.cubes)}")
print(f"Y cube kind: {graph.cubes[1].kind}")  # Y cube is second
print(f"Y cube type: {type(graph.cubes[1].kind).__name__}")
print(f"Number of pipes: {len(graph.pipes)}")

# Validate the graph - this should now pass
graph.validate()
print("Graph validation: PASSED")

# Visualize the block graph
graph.view_as_html()
```
